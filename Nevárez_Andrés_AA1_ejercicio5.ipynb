## 1. Carga de datos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["figure.figsize"]=(9,4); plt.rcParams["axes.grid"]=True

CSV = "ejercicio_fases_analitica_farmacia.csv" # <- usa este nombre exacto
df = pd.read_csv(CSV)

# tipos y fechas
df["fecha_semana"] = pd.to_datetime(df["fecha_semana"], errors="coerce")
df["semana"] = df["semana"].astype(int)

# features útiles
df["tasa_venta"] = df["unidades_vendidas"] / df["stock_inicio_sem"]
df.head()
```

## 2. Analítica descriptiva (¿qué pasó?)

```python
import matplotlib.ticker as mtick

# --- Top productos por semanas sin stock ---
# (renombra el indicador si quieres mantener consistencia en todo el notebook)
df = df.rename(columns={"stockout_sem": "semana_sin_stock"})
top_sin_stock = (
df.groupby("producto")["semana_sin_stock"]
.mean() # proporción de semanas sin stock
.sort_values(ascending=False)
.head(10)
)

# Muestra la serie (formato porcentaje opcional)
display(top_sin_stock.rename("%_semanas_sin_stock").map(lambda x: f"{x*100:.2f}%"))
ax = top_sin_stock.plot(kind="bar")
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0)) # 0.04 -> 4%
plt.title("% de semanas sin stock por producto") 
plt.ylabel("% de semanas sin stock") 
plt.xlabel("Producto")
plt.tight_layout()
plt.show()

# --- Tendencia de unidades vendidas por semana (total) --- 
serie_total = df.groupby("semana")["unidades_vendidas"].sum() 
ax = serie_total.plot(marker="o")
plt.title("Unidades vendidas (total) por semana")
plt.ylabel("Unidades") 
plt.xlabel("Semana") 
plt.tight_layout()
plt.show()
```

## 3. Predictiva (¿qué podría pasar?) + Prescriptiva (¿qué conviene hacer?)

```python
# === PREDICTIVA: tendencia lineal del total y (opcional) por producto ===

# Total por semana
serie = df.groupby("semana")["unidades_vendidas"].sum().reset_index() 
X = serie["semana"].to_numpy().reshape(-1,1)
y = serie["unidades_vendidas"].to_numpy()
Xd = np.hstack([X, np.ones_like(X)]) # [semana, 1]
a,b = (np.linalg.pinv(Xd.T@Xd)@(Xd.T@y)).tolist() # mínimos cuadrados
next_week = int(serie["semana"].max() + 1)
y_next = a*next_week + b
print(f"Pronóstico TOTAL para la semana {next_week}: {y_next:.1f} unidades")

# (Opcional) por producto 
prod = "Paracetamol_500mg" 
s_prod = (df[df["producto"]==prod].groupby("semana")["unidades_vendidas"].sum().reset_index())
if len(s_prod) >= 3:
    Xp = s_prod["semana"].to_numpy().reshape(-1,1) 
    yp = s_prod["unidades_vendidas"].to_numpy() 
    Xpd = np.hstack([Xp, np.ones_like(Xp)])
    ap,bp = (np.linalg.pinv(Xpd.T@Xpd)@(Xpd.T@yp)).tolist()
    y_next_prod = ap*next_week + bp
    print(f"{prod}: pronóstico para semana {next_week} ={y_next_prod:.1f} unidades")


# === PRESCRIPTIVA: alertas de reorden en la última semana observada ===

# Asegurar tasa_venta
if "tasa_venta" not in df.columns:
    df = df.copy()
    df["tasa_venta"] = df["unidades_vendidas"] / df["stock_inicio_sem"].replace(0, np.nan)
last_w = int(df["semana"].max())
base = df.loc[df["semana"] == last_w].copy()
base["tasa_venta"] = base.get("tasa_venta", base["unidades_vendidas"] / base["stock_inicio_sem"].replace(0, np.nan))

# Pronóstico simple por producto = media de últimas 3 semanas
agg = df.groupby(["producto","semana"], as_index=False)["unidades_vendidas"].sum()
hist3 = (agg.sort_values(["producto","semana"]).groupby("producto", group_keys=False).tail(3))
prom3 = (hist3.groupby("producto", as_index=False)["unidades_vendidas"].mean().rename(columns={"unidades_vendidas":"forecast_next"}))
base = base.merge(prom3, on="producto", how="left")
base["forecast_next"] = base["forecast_next"].fillna(base["unidades_vendidas"]).clip(lower=1
                                                                                     )
# Cobertura en semanas
base["cobertura_sem"] = (base["stock_fin_sem"] /base["forecast_next"]).replace([np.inf,-np.inf], np.nan).fillna(999.0)

# Percentiles para lógica C
p25_stock = df.groupby("producto")["stock_fin_sem"].quantile(0.25).rename("p25_stock")
p75_tasa = df.groupby("producto")["tasa_venta"].quantile(0.75).rename("p75_tasa")
base = base.merge(p25_stock, on="producto", how="left").merge(p75_tasa, on="producto", how="left")


# Lógica A — umbrales simples (conservador) 
UMBRAL_COBERTURA_A = 1.5
UMBRAL_LEADTIME_A = 4
UMBRAL_STOCK_BAJO = 30 
base["alerta_A"] = ((base["cobertura_sem"] < UMBRAL_COBERTURA_A) | (base["stock_fin_sem"] <= UMBRAL_STOCK_BAJO) |
((base["tasa_venta"] > 0.50) & (base["lead_time_dias"] >= UMBRAL_LEADTIME_A))).astype(int)

# Lógica B — cobertura + lead time (recomendada) 
UMBRAL_COBERTURA_B = 1.2
UMBRAL_LEADTIME_B = 5 
base["alerta_B"] = ((base["cobertura_sem"] < UMBRAL_COBERTURA_B) & (base["lead_time_dias"] >= UMBRAL_LEADTIME_B)).astype(int)

# Lógica C — percentiles (auto-calibrada)
base["alerta_C"] = ((base["stock_fin_sem"] <= base["p25_stock"].fillna(base["stock_fin_sem"].min())) |
((base["tasa_venta"] >= base["p75_tasa"].fillna(base["tasa_venta"].median())) & (base["lead_time_dias"] >= 5))).astype(int)

# Motivos y priorización
base["motivo_A"] = np.where(base["alerta_A"]==1, "Cobertura/Stock bajo o Tasa+LT", "")
base["motivo_B"] = np.where(base["alerta_B"]==1, "Cobertura<umbral & LeadTime", "")
base["motivo_C"] = np.where(base["alerta_C"]==1, "Percentiles: stock bajo o tasa alta", "")
lead_max = max(1., float(base["lead_time_dias"].max()))
tasa_max = max(1e-9, float(base["tasa_venta"].max())) 
inv_cov = 1.0 / np.maximum(base["cobertura_sem"], 0.1)
base["score_prioridad"] = 0.5*inv_cov + 0.3*(base["lead_time_dias"]/lead_max) + 0.2*(base["tasa_venta"]/tasa_max)
cols = ["producto","stock_fin_sem","forecast_next","cobertura_sem","tasa_venta","lead_time_dias",
"alerta_A","alerta_B","alerta_C","motivo_A","motivo_B","motivo_C","score_prioridad"]
tabla_alertas = base[cols].sort_values( ["alerta_B","alerta_A","alerta_C","score_prioridad"], ascending=[False,False,False,False]).reset_index(drop=True)

print("Resumen de alertas (última semana): "f"A={int(base['alerta_A'].sum())} | B={int(base['alerta_B'].sum())} | C={int(base['alerta_C'].sum())}")
display(tabla_alertas.head(15))
```
